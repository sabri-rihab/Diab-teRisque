```python
# _____________________<( drop the index column )>___________________
df = df.drop(columns=['Unnamed: 0'])
```

```python
# _____________________<( search for outliers )>___________________
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x=df['Pregnancies'])
plt.show()

# note : no outliers here
```

```python
# _____________________<( search for outliers )>___________________
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x=df['Glucose'])
plt.show()
```

```python
import matplotlib.pyplot as plt

sns.scatterplot(x=df['Glucose'], y=df['Age'])
plt.xlabel('Glucose')
plt.ylabel('Age')
plt.title('age vs Glucose')
plt.show()
```

```python
import pandas as pd

df['Glucose'] = df['Glucose'].replace(0, pd.NA)
# 5 null values
Glucose_median = df['Glucose'].median()
print(Glucose_median)
df['Glucose'] = df['Glucose'].fillna(Glucose_median)
```

```python
import matplotlib.pyplot as plt

sns.scatterplot(x=df['Glucose'], y=df['Age'])
plt.xlabel('Glucose')
plt.ylabel('Age')
plt.title('age vs Glucose')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x=df['BloodPressure'])
plt.show()
```

```python
import pandas as pd

df['BloodPressure'] = df['BloodPressure'].replace(0, pd.NA)
df.loc[df['BloodPressure'] < 40, 'BloodPressure'] = pd.NA
# 5 null values
BloodPressure_median = df['BloodPressure'].median()
print(BloodPressure_median)
df['BloodPressure'] = df['BloodPressure'].fillna(BloodPressure_median)
```

```python
import matplotlib.pyplot as plt

sns.scatterplot(x=df['BloodPressure'], y=df['Age'])
plt.xlabel('BloodPressure')
plt.ylabel('Age')
plt.title('age vs BloodPressure')
plt.show()
```
