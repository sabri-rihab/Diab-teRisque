| Metric | What it represents | Simple question |
|---|---|---|
| **Accuracy** | Percentage of all predictions that were correct | How many predictions did the model get right overall? |
| **Precision** | Of the people predicted as **High Risk**, how many were actually High Risk? | When the model says High Risk, how often is it right? |
| **Recall / Sensitivity** | Of the people who are **actually High Risk**, how many did the model detect? | How many of the real High Risk people did we find? |
| **Specificity** | Of the people who are **actually Low Risk**, how many did the model correctly identify? | How many Low Risk people did we correctly recognize? |
| **F1-score** | Balance between Precision and Recall | Can the model both avoid false alarms and detect High Risk cases? |
| **Balanced Accuracy** | Average of Sensitivity and Specificity | How well does the model perform on both classes? |
| **Confusion Matrix** | Shows TP, TN, FP, and FN directly | What types of mistakes is the model making? |

| Metric | Formula | Focus |
|---|---|---|
| **Accuracy** | `(TP + TN) / (TP + TN + FP + FN)` | Overall correctness |
| **Precision** | `TP / (TP + FP)` | Correctness of High Risk predictions |
| **Recall / Sensitivity** | `TP / (TP + FN)` | Finding actual High Risk cases |
| **Specificity** | `TN / (TN + FP)` | Finding actual Low Risk cases |
| **F1-score** | `2 × (Precision × Recall) / (Precision + Recall)` | Precision + Recall balance |
| **Balanced Accuracy** | `(Sensitivity + Specificity) / 2` | Equal importance to both classes |

## <( Pipeline )>

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
import joblib

# X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

pipline = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LogisticRegression(random_state=42,
                                C=1,
                                class_weight=None,
                                penalty='l2',
                                solver='liblinear'
                            ))
])

pipline.fit(X_train, y_train)
y_pred = pipline.predict(X_test)


print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))

joblib.dump(pipline, '../models/logistecR_diabete.joblib')


print("Pipeline sauvegardé dans models/pipeline.joblib  ^^ ")
```

```python
import pandas as pd
from mlflow.models import infer_signature

X_sample = pd.DataFrame(X_train[:5], columns=x.columns)
Y_sample = pipline.predict(X_sample)
signature = infer_signature(X_sample, Y_sample)
```

```python
with mlflow.start_run(run_name="pipeline_registry"):
    mlflow.sklearn.log_model(
        sk_model=pipline,
        artifact_path="pipeline",
        signature=signature,
        registered_model_name="DiabetesRiskPipeline"   # ← nom dans le Registry
    )

print("Pipeline enregistré dans MLflow !")
```
